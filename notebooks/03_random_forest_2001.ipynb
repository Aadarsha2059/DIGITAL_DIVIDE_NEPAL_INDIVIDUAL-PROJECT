```python
import sys
import os
import matplotlib.pyplot as plt

# Add project root to path
sys.path.append(os.path.abspath(".."))

from src.random_forest_model import load_data, prepare_features, train_random_forest

# Load dataset
df = load_data("../data_processed/df_2001.csv")

# Prepare features and target
X, y = prepare_features(df)

# Train Random Forest
rf_model, X_test, y_test, y_pred = train_random_forest(X, y)

# Plot actual vs predicted
plt.figure(figsize=(10,6))
plt.scatter(y_test, y_pred, color='green')
plt.plot([0, 100], [0, 100], 'r--')
plt.xlabel("Actual Electricity Access Rate (%)")
plt.ylabel("Predicted Electricity Access Rate (%)")
plt.title("Random Forest: Actual vs Predicted (2001)")
plt.grid(True)
plt.show()
```
