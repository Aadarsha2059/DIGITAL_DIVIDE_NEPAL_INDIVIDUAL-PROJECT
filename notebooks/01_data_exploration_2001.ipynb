```python
import pandas as pd

# Load CSV files
pop = pd.read_csv("../data_raw/population_2001_zone_corrected.csv")
literacy = pd.read_csv("../data_raw/literacy_2001_zone_corrected.csv")
tech = pd.read_csv("../data_raw/tech_access_2001_zone_corrected.csv")

# Merge datasets
df = pop.merge(literacy, on=["Zone","District"])
df = df.merge(tech, on=["Zone","District"])

# Calculate tech access rates
df['Electricity_Access_Rate'] = df['Households_with_Electricity'] / df['Households_Total'] * 100
df['Radio_Access_Rate'] = df['Households_with_Radio'] / df['Households_Total'] * 100
df['TV_Access_Rate'] = df['Households_with_TV'] / df['Households_Total'] * 100
df['Telephone_Access_Rate'] = df['Households_with_Telephone'] / df['Households_Total'] * 100

# Save cleaned dataset
df.to_csv("../data_processed/df_2001.csv", index=False)
print("Cleaned data saved to data_processed/df_2001.csv")
```
