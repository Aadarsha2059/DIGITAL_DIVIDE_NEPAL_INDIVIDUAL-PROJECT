```python
import sys
import os
import matplotlib.pyplot as plt

# Add project root to path
sys.path.append(os.path.abspath(".."))

from src.xgboost_model import load_data, prepare_features, train_xgboost

# --- Load dataset ---
df = load_data("../data_processed/df_2001.csv")

# --- Prepare features and target ---
X, y = prepare_features(df, target="Electricity_Access_Rate")  # can change target here

# --- Train XGBoost ---
xgb_model, X_test, y_test, y_pred = train_xgboost(X, y)

# --- Plot Actual vs Predicted ---
plt.figure(figsize=(10,6))
plt.scatter(y_test, y_pred, color='orange')
plt.plot([0, 100], [0, 100], 'r--')  # perfect prediction line
plt.xlabel("Actual Electricity Access Rate (%)")
plt.ylabel("Predicted Electricity Access Rate (%)")
plt.title("XGBoost: Actual vs Predicted (2001)")
plt.grid(True)
plt.show()
```
