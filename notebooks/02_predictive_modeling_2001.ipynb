```python
import sys
import os
import matplotlib.pyplot as plt
import pandas as pd

# --- FIX: add project root to sys.path ---
# Assumes your notebook is in DIGITAL_DIVIDE/notebooks/
project_root = os.path.abspath("..")  # one level up from notebooks/
sys.path.append(project_root)

# Now you can import the predictive_model module
from src.predictive_model import load_data, prepare_features, train_predictive_model

# --- Load cleaned dataset ---
df = load_data("../data_processed/df_2001.csv")  # explicitly give path

# --- Prepare features and target ---
X, y = prepare_features(df)

# --- Train predictive model ---
model, X_test, y_test, y_pred = train_predictive_model(X, y)

# --- Plot actual vs predicted ---
plt.figure(figsize=(10,6))
plt.scatter(y_test, y_pred, color='blue')
plt.plot([0, 100], [0, 100], 'r--')  # perfect prediction line
plt.xlabel("Actual Electricity Access Rate (%)")
plt.ylabel("Predicted Electricity Access Rate (%)")
plt.title("Predictive Model: Actual vs Predicted (2001)")
plt.grid(True)

# --- Save figure ---
plt.savefig("../reports/predictions_2001.png", dpi=300)
plt.show()
```
